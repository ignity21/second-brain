# API

```python
def get_stock_data(symbol: List[str], start: Optional[str]=None, 
                   end: Optional[str]=None, period: str="2y", 
                   interval: str="1d", session: Optional[requests.Session]=None):
    if end is None:
        end = date.today().strftime('%Y-%m-%d')
    ticker = yf.Ticker(symbol, session=session)
    data = ticker.history(start=start, end=end, interval=interval, period=period, actions=False)
    data.reset_index(inplace=True)
    data.rename(columns={
        "Date": "date",
        "Open": "open",
        "Low": "low",
        "Close": "close",
        "Adj Close": "adj_close",
        "Volume": "volume"
    }, inplace=True)
    data['symbol'] = symbol
    return data
```

```python
stock_data = get_stock_data("AAPL", start="2022-01-01", session=session)
stock_data.head()
```
