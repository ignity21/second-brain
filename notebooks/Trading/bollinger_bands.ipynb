```python
import pandas as pd
import akshare as ak
```

```python
df = ak.fund_etf_hist_sina(symbol="sh510050")
df.date = pd.to_datetime(df.date)
df = df.set_index("date")
df = df["2018-1-1":]
df.tail()
```

```python
window_size = 30
std_factor = 1.5

rolling_close = df.close.rolling(window_size)
sma30 = rolling_close.mean()
std30 = rolling_close.std()
deviation = std_factor * std30
upperband = sma30 + deviation
lowerband = sma30 - deviation
```

```python
df.close.plot(figsize=(20, 10))
sma30.plot()
upperband.plot()
lowerband.plot()
```
