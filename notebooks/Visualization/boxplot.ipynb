# [Boxplot](https://www.data-to-viz.com/caveat/boxplot.html)

- The difference between Quartiles 1 and 3 is called the interquartile range (IQR)
- The extreme line shows Q3+1.5xIQR to Q1-1.5xIQR (the highest and lowest value excluding outliers).
- Dots (or other markers) beyond the extreme line shows potntial outliers.

![](images/boxplot.png)

```python
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
df.head()
```

## seaborn

```python
fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(16, 12))

# basic boxplot
sns.boxplot(x=df["species"], y=df["sepal_length"], ax=axs[0][0])
axs[0][0].title.set_text("basic boxplot")

# boxplot with jitter
sns.boxplot(x=df["species"], y=df["sepal_length"], ax=axs[0][1]) 
sns.stripplot(x=df["species"], y=df["sepal_length"], ax=axs[0][1], color="gray", jitter=0.2, size=5)
axs[0][1].title.set_text("boxplot with jitter")

sns.violinplot(x=df["species"], y=df["sepal_length"], ax=axs[1][0]) # violinplot
axs[1][0].title.set_text("violinplot")
```
