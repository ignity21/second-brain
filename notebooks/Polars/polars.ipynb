```python
import polars as pl
from rich import print
```

```python
sales = pl.read_csv("practice_sales.csv")
print(sales.shape)
print(sales.schema) # column names and dtype
print(sales.glimpse()) # useful if we have many columns
sales.head(3)
```

# Selection
- By Rows: `df[0]`, `df[-1]`, `df[1:3]`
- By Columns: `df["region"]`, `df[["region", "channel"]]`
- By Both (row first, then columns): `df[:3, ["region", "channel"]]`
- `.select()`: `df.select("region", "channel")`

# Sorting
- `df.sort("quantity", descending=True)` # default ascending
- By multiple columns: `df.sort("quantity", "product")`

# Summarizing
- Finding extreme values: `df.top_k(3, by="quantity")` or `bottom_k`
- `df.describe()`
